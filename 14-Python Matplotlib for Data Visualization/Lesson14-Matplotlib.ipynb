# 1. Import Required Library

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 1.1 Read CSV File

df = pd.read_csv("sales.csv")
df.head()

## 1.2 Line Plot

```python
plt.plot(df["Date"], df["Total_Sales"], marker="o", color="blue", label="Sales")
plt.title("Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales ($)")
plt.gcf().autofmt_xdate()
plt.legend()
plt.tight_layout()
plt.show()
```

## 1.3 Bar Plot

```python
category_profit = df.groupby("Product_Category")["Profit"].sum().sort_values(ascending=False)
bars = plt.bar(category_profit.index, category_profit.values, color="green" ,width=.5)
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        yval + 50,
        f"${yval:,.0f}",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold",
    )
plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit ($)")
plt.show()
```

## 1.4 Pie Chart

```python
category_sales = df.groupby("Region")["Total_Sales"].sum()
plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
)
plt.title("Sales Distribution")
plt.tight_layout()
plt.show()
```

## 1.5 Scatter Plot

```python
scatter = plt.scatter(
    df["Total_Sales"],
    df["Profit"],
    s=df["Quantity"] * 30,
    c=df["Quantity"],
    cmap="coolwarm",
    alpha=0.9,
    edgecolors="black",
)
plt.title("Correlation: Sales vs Profit (Size = Quantity)", fontsize=12, weight="bold")
plt.xlabel("Total Sales ($)")
plt.ylabel("Profit ($)")
cbar = plt.colorbar(scatter)
cbar.set_label("Quantity Purchased")
plt.tight_layout()
plt.show()
```

## 1.6 Box Plot

```python
plt.boxplot(df["Total_Sales"])
plt.title("Sales Outliers Detection")
plt.ylabel("Sales ($)")
plt.xticks([1], ["All Transactions"])
plt.tight_layout()
plt.show()
```

## 1.7 Histogram

```python
plt.hist(df["Total_Sales"], color="skyblue", edgecolor="black")
plt.title("Frequency Distribution of Sales Values")
plt.xlabel("Sales Ranges ($)")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```
