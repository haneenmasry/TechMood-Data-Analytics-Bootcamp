# مشروع تنظيف ومعالجة بيانات المبيعات (Sales Data Cleaning & Manipulation)

يحتوي هذا الملف على الخطوات الكاملة لإنشاء البيانات، فحصها، وتطبيق عمليات التنظيف والتحليل المطلوبة باستخدام مكتبة **Pandas** خطوة بخطوة.

### 1. استيراد المكتبات وبناء الـ DataFrame والفحص الأولي
نقوم باستيراد المكتبات الأساسية وتوليد البيانات لفحصها باستخدام `head()` و `info()` و `describe()`.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Product': ['iPhone', 'Samsung S23', 'iPhone', 'MacBook', 'Samsung S23', 'iPad', 'MacBook', 'iPhone'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Computers', 'Electronics', 'Electronics', 'Computers', 'Electronics'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Sales': [1200, 950, 1200, 2500, None, 600, 2500, 1200], 
    'Quantity': [2, 1, 2, 1, 3, np.nan, 2, 1],
    'OrderDate': ['2026-01-15', '2026-01-16', '2026-01-16', '2026-01-17', '2026-01-18', '2026-01-19', '2026-01-20', '2026-01-20']
}

df = pd.DataFrame(data)
```

```python
# عرض أول 5 صفوف
df.head()
```

```python
# فحص جودة وأنواع البيانات والتعرف على القيم المفقودة
df.info()
```

```python
# الإحصاءات الوصفية للبيانات الرقمية
df.describe()
```

### 2. معالجة وتصحيح أنواع البيانات (Data Types)
تحويل الأعمدة إلى أنواع البيانات المناسبة لكل منها لضمان دقة العمليات الحسابية والزمنية اللاحقة.

```python
df['Quantity'] = df['Quantity'].astype('Int32')
df['OrderDate'] = df['OrderDate'].astype('datetime64[us]')
print(df.dtypes)
```

### 3. اكتشاف ومعالجة القيم المفقودة (Missing Values)
التحقق من مواضع الفراغات ومعالجتها عن طريق استبدالها بقيم منطقية أو ثابتة.

```python
# التحقق من وجود القيم المفقودة
df.isnull().sum()
```

```python
# معالجة القيم المفقودة
df['Quantity'] = df['Quantity'].fillna(6)
df['Sales'] = df['Sales'].fillna(800).astype('float64')
df
```

### 4. تطبيق التصفية (Filtering)
استخراج البيانات بناءً على شروط مخصصة مدمجة باستخدام معامل التصفية المنطقي (OR).

```python
custom_filter_df = df[(df['Region'] == 'North') | (df['Product'] == 'iPhone') | (df['Sales'] > 1000)]
print(custom_filter_df)
```

### 5. تجميع البيانات وتحليلها باستخدام `groupby()`
حساب إجمالي المبيعات ومتوسط المبيعات لكل فئة من المنتجات مع تنظيم أسماء الأعمدة الناتجة.

```python
category_summary = df.groupby('Category')['Sales'].agg(['sum', 'mean'])
category_summary = category_summary.rename(columns={
    'sum': 'Total_Sales',
    'mean': 'Average_Sales'
})
print(category_summary)
```

### 6. إتمام عمليات التنظيف وحفظ النسخة النهائية

```python
df.to_csv('final_cleaned_sales_data.csv', index=False)
print("تم حفظ النسخة النهائية بنجاح !")
```
